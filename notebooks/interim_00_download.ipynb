Both sources are public; no auth / API key needed. Downloading is resumable at the zip granularity for RAVDESS (already-extracted actors are skipped). For CREMA-D the repo is cloned once and then media are pulled via Git LFS; a retried run will no-op once files are present.

```python
from __future__ import annotations

import sys
from pathlib import Path

ROOT = Path.cwd().resolve()
while not (ROOT / "src").is_dir() and ROOT != ROOT.parent:
    ROOT = ROOT.parent
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))

DATA = ROOT / "data"
DATA.mkdir(parents=True, exist_ok=True)
print("downloads will land under", DATA)
```
